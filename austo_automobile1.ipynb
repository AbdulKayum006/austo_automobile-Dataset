```python
# Explore dataset with specific column types
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# File path
filename = '/content/austo_automobile1.csv'

# Column names
names = [
    'Age', 'Gender', 'Profession', 'Marital_status', 'Education',
    'No_of_Dependents','Personal_loan', 'House_loan', 'Partner_working',
    'Salary', 'Partner_salary', 'Total_salary', 'Price', 'Make'
]

# Load dataset with column names
dataset = pd.read_csv(filename, names=names)

# Show first 20 rows
print(dataset.head(20))
```

```python
# Check the structure of the dataset

# 1. Basic info: column names, non-null counts, dtypes
print("\n--- Dataset Info ---")
print(dataset.info())

# 2. Shape (rows, columns)
print("\n--- Dataset Shape ---")
print(dataset.shape)

# 3. First 5 rows (quick preview)
print("\n--- First 5 Rows ---")
print(dataset.head())

# 4. Summary of numerical columns
print("\n--- Statistical Summary ---")
print(dataset.describe())

# 5. Check missing values per column
print("\n--- Missing Values ---")
print(dataset.isnull().sum())
```

```python
# Check data types of each column
print("\n--- Data Types of Each Column ---")
print(dataset.dtypes)
```

```python
# Columns to convert
numeric_columns = ['Age', 'No_of_Dependents', 'Salary', 'Partner_salary', 'Total_salary', 'Price']

# Convert columns to numeric, coerce errors to NaN if any invalid values
for col in numeric_columns:
    dataset[col] = pd.to_numeric(dataset[col], errors='coerce')

# Check data types after conversion
print("\n--- Data Types After Conversion ---")
print(dataset.dtypes)
```

```python
# Convert 'Education' column to object (string) type
dataset['Education'] = dataset['Education'].astype('object')

# Check data types after conversion
print("\n--- Data Types After Conversion ---")
print(dataset.dtypes)
```

```python
# Check missing values for each column
missing_values = dataset.isnull().sum()

print("\n--- Missing Values in Each Column ---")
print(missing_values)
```

```python
# Separate numeric and categorical columns
numeric_columns = ['Age', 'No_of_Dependents', 'Salary', 'Partner_salary', 'Total_salary', 'Price']
categorical_columns = ['Gender', 'Profession', 'Marital_status', 'Education', 'House_loan', 'Partner_working', 'Make']

# Fill missing values for numeric columns with median
for col in numeric_columns:
    median_value = dataset[col].median()
    dataset[col].fillna(median_value, inplace=True)

# Fill missing values for categorical columns with mode
for col in categorical_columns:
    mode_value = dataset[col].mode()[0]
    dataset[col].fillna(mode_value, inplace=True)

# Verify missing values are treated
print("\n--- Missing Values After Treatment ---")
print(dataset.isnull().sum())
```

```python
# Show first 50 rows of the dataset
print(dataset.head(50))
```

```python
# Statistical summary for numeric columns
print("\n--- Numeric Columns Summary ---")
print(dataset.describe())

# Statistical summary for categorical columns
print("\n--- Categorical Columns Summary ---")
print(dataset.describe(include='object'))
```

```python
import pandas as pd

# 1. Check for outliers using basic statistics
print("\n--- Outliers Check (Numeric Columns) ---")
for col in ['Age', 'No_of_Dependents', 'Salary', 'Partner_salary', 'Total_salary', 'Price']:
    q1 = dataset[col].quantile(0.25)
    q3 = dataset[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = dataset[(dataset[col] < lower_bound) | (dataset[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers")

# 2. Check for invalid or inconsistent categories
print("\n--- Categorical Columns Check ---")
for col in ['Gender', 'Profession', 'Marital_status', 'Education', 'House_loan', 'Partner_working', 'Make']:
    unique_vals = dataset[col].unique()
    print(f"{col}: {unique_vals}")

# 3. Check for negative or zero values in numeric columns where not allowed
print("\n--- Negative or Zero Values Check ---")
for col in ['Age', 'Salary', 'Partner_salary', 'Total_salary', 'Price']:
    invalid_values = dataset[dataset[col] <= 0]
    print(f"{col}: {len(invalid_values)} invalid values")
```

```python
# 1. Treat outliers in numeric columns using IQR capping
numeric_columns = ['Age', 'No_of_Dependents', 'Salary', 'Partner_salary', 'Total_salary', 'Price']

for col in numeric_columns:
    q1 = dataset[col].quantile(0.25)
    q3 = dataset[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    # Cap the outliers
    dataset[col] = dataset[col].apply(lambda x: lower_bound if x < lower_bound else (upper_bound if x > upper_bound else x))

# 2. Standardize categorical values
dataset['Gender'] = dataset['Gender'].str.capitalize()  # 'male' → 'Male', 'female' → 'Female'
dataset['Marital_status'] = dataset['Marital_status'].str.capitalize()
dataset['Education'] = dataset['Education'].str.capitalize()
dataset['House_loan'] = dataset['House_loan'].str.capitalize()
dataset['Partner_working'] = dataset['Partner_working'].str.capitalize()

# 3. Replace negative or zero values in numeric columns with median
for col in ['Age', 'Salary', 'Partner_salary', 'Total_salary', 'Price']:
    median_value = dataset[col].median()
    dataset[col] = dataset[col].apply(lambda x: median_value if x <= 0 else x)

# Verify changes
print("\n--- Dataset after treating irregularities ---")
print(dataset.head())
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Standardize 'Make' column for consistency
dataset['Make'] = dataset['Make'].str.upper()  # e.g., 'SUV', 'SEDAN', etc.
dataset['Gender'] = dataset['Gender'].str.capitalize()  # 'male', 'female' -> 'Male', 'Female'

# Filter salaried males (Salary > 0 and Gender == 'Male')
salaried_males = dataset[(dataset['Salary'] > 0) & (dataset['Gender'] == 'Male')]

# Count how many buy SUVs and Sedans
suv_count = salaried_males[salaried_males['Make'].str.contains('SUV')].shape[0]
sedan_count = salaried_males[salaried_males['Make'].str.contains('SEDAN')].shape[0]
total = salaried_males.shape[0]

# Calculate proportions
suv_likelihood = suv_count / total
sedan_likelihood = sedan_count / total

print(f"Total salaried males: {total}")
print(f"SUV buyers: {suv_count} ({suv_likelihood*100:.2f}%)")
print(f"Sedan buyers: {sedan_count} ({sedan_likelihood*100:.2f}%)")

# Visualization
plt.figure(figsize=(6,4))
sns.barplot(x=['SUV', 'Sedan'], y=[suv_likelihood*100, sedan_likelihood*100], palette='pastel')
plt.ylabel('Percentage of Salaried Male Buyers (%)')
plt.title('Salaried Male Preference: SUV vs Sedan')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Standardize 'Partner_working' column
dataset['Partner_working'] = dataset['Partner_working'].str.capitalize()  # 'yes' → 'Yes', 'no' → 'No'

# --- 1. Summary statistics ---
price_by_partner = dataset.groupby('Partner_working')['Price'].describe()
print("\n--- Price Summary by Partner Working Status ---")
print(price_by_partner)

# --- 2. Mean and Median Price ---
mean_price = dataset.groupby('Partner_working')['Price'].mean()
median_price = dataset.groupby('Partner_working')['Price'].median()

print("\n--- Mean Price by Partner Working Status ---")
print(mean_price)
print("\n--- Median Price by Partner Working Status ---")
print(median_price)

# --- 3. Boxplot to compare distributions ---
plt.figure(figsize=(6,4))
sns.boxplot(x='Partner_working', y='Price', data=dataset, palette='pastel')
plt.title('Automobile Price by Partner Working Status')
plt.ylabel('Price')
plt.show()

# --- 4. Proportion of high-priced cars (top 25%) ---
q75 = dataset['Price'].quantile(0.75)
dataset['High_Price'] = dataset['Price'] >= q75

high_price_prop = dataset.groupby('Partner_working')['High_Price'].mean() * 100
print("\n--- % of High-Priced Cars by Partner Working Status ---")
print(high_price_prop)

# Barplot for high-priced proportion
plt.figure(figsize=(6,4))
sns.barplot(x=high_price_prop.index, y=high_price_prop.values, palette=['lightblue','lightgreen'])
plt.title('Proportion of High-Priced Cars by Partner Working Status')
plt.ylabel('Percentage (%)')
plt.show()
```
